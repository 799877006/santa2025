```python
import pandas as pd
from decimal import Decimal,getcontext
getcontext().prec = 50
df = pd.read_csv('70.66.csv')
```

```python
for col in ['x', 'y', 'deg']:
    df[col] = df[col].astype(str).str.strip('s')
df[['group_id', 'item_id']] = df['id'].str.split('_', expand=True)
for group in df['group_id'].unique():
    if group == '001':
        side_length = Decimal('0.8')    
        square_size = Decimal('0.064')
        print("side_length", side_length)
        print("square_size", square_size)
    else:
        max_x = Decimal(df[df['group_id'] == group]['x'].max())
        max_y = Decimal(df[df['group_id'] == group]['y'].max())
        min_x = Decimal(df[df['group_id'] == group]['x'].min())
        min_y = Decimal(df[df['group_id'] == group]['y'].min())
        side_length = max(max_x - min_x, max_y - min_y)
        print("side_length", side_length)
        print("square_size", side_length ** 2)
        
    
```
